### 格点上实际算什么？

```text
规范场组态
    ↓
核子两点函数 C2：核子从源传播到汇
    ↓ 加入胶子 EMT 算符
核子三点函数 C3：传播过程中探测胶子结构
    ↓
比值 R = C3/C2
    ↓
提取基态矩阵元
    ↓
重整化与 matching
```


$$
C_2(\mathbf p,t_s)
=\sum_{\mathbf x}e^{-i\mathbf p\cdot\mathbf x}
\operatorname{Tr}\!\left[
\Gamma\,\langle\chi_N(\mathbf x,t_s)\bar\chi_N(\mathbf0,0)\rangle
\right].
$$

$$
\begin{aligned}
C_{3,g}^{\mu\nu}(\mathbf p',\mathbf p;t_s,\tau;t_f)
={}&\sum_{\mathbf x,\mathbf y}
 e^{-i\mathbf p'\cdot\mathbf x}
 e^{+i(\mathbf p'-\mathbf p)\cdot\mathbf y}
\\[-2pt]
&\times\operatorname{Tr}\!\left[
\Gamma\,\left\langle
\chi_N(\mathbf x,t_s)
\Delta T_g^{\mu\nu}(\mathbf y,\tau;t_f)
\bar\chi_N(\mathbf0,0)
\right\rangle\right],
\end{aligned}
$$
